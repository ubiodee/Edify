```python
!pip install openai
from openai import OpenAI
client = OpenAI(api_key="")

# Function to generate business report from user inputs
def generate_business_report(employee_responses):
    response = client.chat.completions.create(
        model="gpt-4o",
        messages=[
            {
                "role": "system",
                "content": """
                You are Edify business coach assistant. You will receive the responses from different employees about what they think of their company. Analyze these responses, both from top management and employees, identifying insights based on each employee's unique feedback. You will perform a General Audit of their responses and offer actionable advice using categories like Purpose, Vision, Operations, Culture, and Communication, but only use relevant sections. End with detailed recommendations and insights.
                """
            },
            {
                "role": "user",
                "content": employee_responses  # The input provided dynamically by the user
            }
        ],
        temperature=0.87,
        max_tokens=2150,
        top_p=1,
        frequency_penalty=0.64,
        presence_penalty=0
    )

    # Extract and return the assistant's response
    # Access the 'choices' attribute first, then access elements within the list
    return response.choices[0].message.content


```
